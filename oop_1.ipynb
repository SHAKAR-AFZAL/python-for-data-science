```python
# Python Challenge (OOP Concept Encapsulation)
#===============================================
#       1.Bank Account Management System
class BankAccount:
    def __init__(self, account_number, owner, balance):
        self.account_number = account_number
        self.owner = owner
        self.__balance = balance

    # Getter
    def get_balance(self):
        return self.__balance

    # Deposit
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Invalid amount")

    # Withdraw
    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")

    def display_account(self):
        print("\n--- Account Information ---")
        print("Account:", self.account_number)
        print("Owner:", self.owner)
        print("Balance:", self.__balance)


account = BankAccount("PK001", "Mian Shakar Afzal", 50000)
account.deposit(10000)
account.withdraw(15000)
account.display_account()
```

```python
#=====================================
# 2. Employee Payroll System
#=====================================
class Employee:
    def __init__(self, employee_id, name, salary):
        self.employee_id = employee_id
        self.name = name
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary
        else:
            print("Salary must be greater than 0")

    def calculate_bonus(self, percentage):
        return self.__salary * percentage / 100

    def calculate_tax(self, percentage):
        return self.__salary * percentage / 100

    def calculate_net_salary(self, bonus_percentage, tax_percentage):

        bonus = self.calculate_bonus(bonus_percentage)
        tax = self.calculate_tax(tax_percentage)

        return self.__salary + bonus - tax


employee = Employee("E001", "Mian Shakar Afzal", 100000)

bonus = employee.calculate_bonus(10)
tax = employee.calculate_tax(5)
net_salary = employee.calculate_net_salary(10, 5)

print("Employee:", employee.name)
print("Basic Salary:", employee.get_salary())
print("Bonus:", bonus)
print("Tax:", tax)
print("Net Salary:", net_salary)
```

```python
#===================================
#3. Inventory Management System
#===================================
class Product:
    def __init__(self, product_id, name, price, quantity):
        self.product_id = product_id
        self.name = name
        self.__price = price
        self.__quantity = quantity

    def get_price(self):
        return self.__price

    def get_quantity(self):
        return self.__quantity

    def set_price(self, price):
        if price > 0:
            self.__price = price

    def add_stock(self, quantity):
        if quantity > 0:
            self.__quantity += quantity

    def sell(self, quantity):

        if quantity <= 0:
            print("Invalid quantity")

        elif quantity > self.__quantity:
            print("Not enough stock")

        else:
            self.__quantity -= quantity
            print(f"{quantity} units sold")

    def inventory_value(self):
        return self.__price * self.__quantity

    def display(self):
        print(
            self.product_id,
            self.name,
            self.__price,
            self.__quantity
        )


laptop = Product("P001", "Laptop", 120000, 10)

laptop.sell(3)
laptop.add_stock(5)

laptop.display()

print("Inventory Value:", laptop.inventory_value())
```

```python
#====================================
#         4. ATM System
#====================================
class ATM:
    def __init__(self, pin, balance):
        self.__pin = pin
        self.__balance = balance

    def verify_pin(self, entered_pin):
        return entered_pin == self.__pin

    def check_balance(self):
        return self.__balance

    def deposit(self, amount):

        if amount > 0:
            self.__balance += amount
            print("Deposit successful")
        else:
            print("Invalid amount")

    def withdraw(self, amount):

        if amount <= 0:
            print("Invalid amount")

        elif amount > self.__balance:
            print("Insufficient balance")

        else:
            self.__balance -= amount
            print("Withdrawal successful")

    @staticmethod
    def welcome_message():
        print("Welcome to Secure ATM")


ATM.welcome_message()
account = ATM(1234, 50000)
entered_pin = int(input("Enter PIN: "))
if account.verify_pin(entered_pin):

    while True:

        print("\n1. Check Balance")
        print("2. Deposit")
        print("3. Withdraw")
        print("4. Exit")
        choice = input("Choose option: ")

        if choice == "1":
            print("Balance:", account.check_balance())

        elif choice == "2":
            amount = float(input("Enter amount: "))
            account.deposit(amount)

        elif choice == "3":
            amount = float(input("Enter amount: "))
            account.withdraw(amount)

        elif choice == "4":
            print("Thank you")
            break

        else:
            print("Invalid option")

else:
    print("Incorrect PIN")
```

```python
#====================================
#     5. Online Shopping Cart
#====================================
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price

class ShoppingCart:

    def __init__(self):
        self.__items = []

    def add_product(self, product):
        self.__items.append(product)
        print(f"{product.name} added to cart")

    def remove_product(self, product):
        if product in self.__items:
            self.__items.remove(product)
            print(f"{product.name} removed")

    def calculate_total(self):

        total = 0

        for product in self.__items:
            total += product.price

        return total

    def display_cart(self):

        print("\n--- Shopping Cart ---")

        for product in self.__items:
            print(product.name, "-", product.price)

        print("Total:", self.calculate_total())

    @staticmethod
    def calculate_discount(total, percentage):

        return total * percentage / 100


laptop = Product("Laptop", 120000)
mouse = Product("Mouse", 2500)
keyboard = Product("Keyboard", 5000)

cart = ShoppingCart()

cart.add_product(laptop)
cart.add_product(mouse)
cart.add_product(keyboard)

cart.display_cart()
total = cart.calculate_total()
discount = ShoppingCart.calculate_discount(total, 10)
final_price = total - discount
print("Discount:", discount)
print("Final Price:", final_price)
```
