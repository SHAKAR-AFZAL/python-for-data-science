```python
# Sum of Numbers
n = int(input("Enter a positive integer: "))
total = 0
for i in range(1, n + 1):
    total = total + i
print("Sum:", total)
```

```python
# Sales Data Total
total_sales = 0
for i in range(1, 6):
    sales = float(input("Enter sales for product " + str(i) + ": "))
    total_sales = total_sales + sales
print("Total Sales:", total_sales)
```

```python
# Data Analysis: Count Positive & Negative Values1
n = int(input("How many numbers? "))
positive = 0
negative = 0
for i in range(1, n + 1):
    number = float(input("Enter number " + str(i) + ": "))
    if number > 0:
        positive = positive + 1
    elif number < 0:
        negative = negative + 1
print("Positive numbers:", positive)
print("Negative numbers:", negative)
```

```python
n = int(input("How many transactions? "))
total_sales = 0
highest_sale = 0
lowest_sale = 0
for i in range(1, n + 1):
    sale = float(input("Enter sale " + str(i) + ": "))
    total_sales = total_sales + sale

    if i == 1:
        highest_sale = sale
        lowest_sale = sale
    else:
        if sale > highest_sale:
            highest_sale = sale

        if sale < lowest_sale:
            lowest_sale = sale

average_sale = total_sales / n

print("Total Sales:", total_sales)
print("Average Sale:", average_sale)
print("Highest Sale:", highest_sale)
print("Lowest Sale:", lowest_sale)
```
