```python
# PYTHON CHALLENGE (IF ELSE STATEMENTS)

# IF MARKS >= 50 Pass Else Fail
marks=int(input("Enter the marks: "))
if marks>=50:
    print("Pass")
else:
    print("Fail")
```

```python
# Even or Odd
a=int(input("Enter the number: "))
if a%2==0:
    print("Even")
else:
    print("Odd")
```

```python
# ATM Withdrawal
current_balance = int(input("Enter the current balance: "))
withdrawal_amount = int(input("Enter the withdrawal amount: "))
if withdrawal_amount <= current_balance:
    print("Withdrawal Successful")
else:
    print("Insufficient Balance")
```

```python
# E-Commerce Free Shipping
order_amount=int(input("Enter the order amount: "))
customer_membership=input("Enter the customer membership status (yes or no): ")
if order_amount>=5000 or customer_membership=="yes":
    print("Shipping is free")
else:
    print("Shipping costs 500")
```

```python
# Employee Bouns
Employee_salary=int(input("Enter the salary: "))
Years_of_experience=int(input("Enter the years of experience: "))
if Employee_salary>=50000 and Years_of_experience>=2:
    print("Eligible for bonus")
else:
    print("Not eligible for bonus")
```
