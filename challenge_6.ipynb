```python
# Industry-Style Function
def calculate_profit():
    selling_price = float(input("Enter selling price: "))
    cost_price = float(input("Enter cost price: "))
    quantity_sold = int(input("Enter quantity: "))

    revenue = selling_price * quantity_sold
    total_cost = cost_price * quantity_sold
    profit = revenue - total_cost
    profit_per_unit = profit / quantity_sold

    print("\nRevenue:", revenue)
    print("Total Cost:", total_cost)
    print("Profit:", profit)
    print("Profit per Unit:", profit_per_unit)


calculate_profit()
```

```python
# map, filter, reduce

from functools import reduce
transactions = [1200, 450, 3000, 800, 2500, 150, 4200, 900]
# 1. MAP → Apply 10% discount to every transaction
discounted_sales = list(
    map(lambda amount: amount * 0.90, transactions)
)

# 2. FILTER → Keep high-value transactions
high_value_sales = list(
    filter(lambda amount: amount >= 1000, discounted_sales)
)

# 3. REDUCE → Calculate total high-value sales
total_revenue = reduce(
    lambda a, b: a + b,
    high_value_sales
)

print("Original Transactions:", transactions)
print("After 10% Discount:", discounted_sales)
print("High-Value Transactions:", high_value_sales)
print("Total High-Value Revenue:", total_revenue)
```
