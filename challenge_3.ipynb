```python
# Python Challenge (TAKING INPUT FROM USER AND PERFORMING SPECIFIC OPERATION)

# BASIC CALCULATOR
a = int(input("ENTER THE FIRST NUMBER:"))
b = int(input("ENTER THE SECOND NUMBER:"))
Sum = int(a)+int(b)
Difference = int(a)-int(b)
Product = int(a)*int(b)
Division = int(a)/int(b)
print("SUM:",Sum)
print("DIFFERENC:",Difference)
print("PRODUCT:",Product)
print("DIVISION:",Division)
```

```python
# Student Marks Calculator
a = float(input("ENTER PROGRAMMING MARKS:"))
b = float(input("ENTER DATABASE MARKS:"))
c = float(input("ENTER MATHEMATICS MARKS:"))
Total = a+b+c
print("TOTAL MARKS:",Total)
Average_Marks = Total / 3
print("Average_Marks:", Average_Marks)
```

```python
# Salary Calculator
salary = float(input("Enter basic salary: "))
bonus = float(input("Enter bonus: "))
allowance = float(input("Enter allowance: "))
total_salary = salary + bonus + allowance
print("Total Salary:", total_salary)
```

```python
# E-Commerce Sales Calculation
price = float(input("Enter product price: "))
quantity = int(input("Enter quantity: "))
discount = float(input("Enter discount amount: "))
subtotal = price * quantity
final_amount = subtotal - discount
print("Subtotal:", subtotal)
print("Final Amount:", final_amount)
```

```python
# Small Project
selling_price = float(input("Enter product selling price: "))
cost_price = float(input("Enter product cost price: "))
quantity = int(input("Enter quantity sold: "))
revenue = selling_price * quantity
total = cost_price * quantity
profit = revenue - total
profit_per_unit = profit / quantity

print("Revenue:", revenue)
print("Total Cost:", total)
print("Profit:", profit)
print("Profit per Unit:", profit_per_unit)
```
