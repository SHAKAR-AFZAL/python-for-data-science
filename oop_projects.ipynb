```python
# OOP PROJECTS
# 1 Banking Management System
from abc import ABC, abstractmethod
class BankAccount(ABC):

    bank_name = "HBL Bank"

    def __init__(self, account_no, owner, balance):
        self.account_no = account_no
        self.owner = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Invalid amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print(f"Withdrawn: {amount}")

    @abstractmethod
    def calculate_interest(self):
        pass

    @staticmethod
    def bank_info():
        print("ABC Bank - Secure Banking System")


class SavingsAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 0.05


class CurrentAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 0.02


BankAccount.bank_info()

account1 = SavingsAccount("SA001", "Mian", 100000)
account2 = CurrentAccount("CA001", "Ali", 100000)

account1.deposit(10000)
account1.withdraw(5000)

print("Mian Balance:", account1.get_balance())
print("Mian Interest:", account1.calculate_interest())

print("Ali Interest:", account2.calculate_interest())
```

```python
# 2 E-Commerce System
from abc import ABC, abstractmethod
class Product:

    def __init__(self, product_id, name, price):
        self.product_id = product_id
        self.name = name
        self.price = price

class ShoppingCart:

    def __init__(self):
        self.__products = []

    def add_product(self, product):
        self.__products.append(product)

    def remove_product(self, product):
        if product in self.__products:
            self.__products.remove(product)

    def calculate_total(self):
        total = 0

        for product in self.__products:
            total += product.price

        return total

    def show_cart(self):
        for product in self.__products:
            print(product.name, product.price)


class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):

    def pay(self, amount):
        print(f"Paid {amount} using Credit Card")

class BankPayment(Payment):

    def pay(self, amount):
        print(f"Paid {amount} using Bank Transfer")


laptop = Product("P01", "Laptop", 120000)
mouse = Product("P02", "Mouse", 2500)

cart = ShoppingCart()

cart.add_product(laptop)
cart.add_product(mouse)

cart.show_cart()

total = cart.calculate_total()

print("Total:", total)

payment = CreditCardPayment()
payment.pay(total)
```

```python
# 3 Employee Management System
from abc import ABC, abstractmethod

class Employee(ABC):

    company = "Tech Solutions"

    def __init__(self, employee_id, name, salary):
        self.employee_id = employee_id
        self.name = name
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary

    @abstractmethod
    def calculate_bonus(self):
        pass

    def display(self):
        print(
            self.employee_id,
            self.name,
            self.__salary
        )


class Developer(Employee):

    def calculate_bonus(self):
        return self.get_salary() * 0.10


class Manager(Employee):

    def calculate_bonus(self):
        return self.get_salary() * 0.20


class DataAnalyst(Employee):

    def calculate_bonus(self):
        return self.get_salary() * 0.15


employees = [
    Developer("E01", "Mian", 100000),
    Manager("E02", "Ahmed", 150000),
    DataAnalyst("E03", "Ali", 120000)
]

for employee in employees:

    employee.display()
    bonus = employee.calculate_bonus()
    print("Bonus:", bonus)
    print()
```

```python
# 4 Ride Booking System
from abc import ABC, abstractmethod

class Vehicle(ABC):

    def __init__(self, vehicle_number):
        self.vehicle_number = vehicle_number

    @abstractmethod
    def calculate_fare(self, distance):
        pass


class Car(Vehicle):

    def calculate_fare(self, distance):
        return distance * 100


class Bike(Vehicle):

    def calculate_fare(self, distance):
        return distance * 50


class Driver:

    def __init__(self, name, vehicle):
        self.name = name
        self.vehicle = vehicle


class Ride:

    def __init__(self, driver, distance):
        self.driver = driver
        self.distance = distance

    def calculate_fare(self):
        return self.driver.vehicle.calculate_fare(
            self.distance
        )

    def show_ride(self):
        print("Driver:", self.driver.name)
        print("Vehicle:", self.driver.vehicle.vehicle_number)
        print("Distance:", self.distance)
        print("Fare:", self.calculate_fare())


car = Car("ABC-1234")
driver = Driver("Ali", car)
ride = Ride(driver, 15)
ride.show_ride()
```

```python
# Inventory & Order Management System
class Product:

    def __init__(self, product_id, name, price, stock):
        self.product_id = product_id
        self.name = name
        self.__price = price
        self.__stock = stock

    def get_price(self):
        return self.__price

    def get_stock(self):
        return self.__stock

    def reduce_stock(self, quantity):

        if quantity <= 0:
            return False

        if quantity > self.__stock:
            return False

        self.__stock -= quantity

        return True

    def display(self):
        print(
            self.product_id,
            self.name,
            self.__price,
            self.__stock
        )

class Order:

    order_count = 0
    def __init__(self, customer):
        self.customer = customer
        self.__items = []

        Order.order_count += 1
        self.order_id = Order.order_count

    def add_product(self, product, quantity):

        if product.reduce_stock(quantity):

            self.__items.append(
                (product, quantity)
            )

            print(
                f"{quantity} x {product.name} added"
            )

        else:
            print("Insufficient stock")

    def calculate_total(self):

        total = 0

        for product, quantity in self.__items:
            total += product.get_price() * quantity

        return total

    def display_order(self):

        print("\nOrder ID:", self.order_id)
        print("Customer:", self.customer)

        for product, quantity in self.__items:

            print(
                product.name,
                "x",
                quantity,
                "=",
                product.get_price() * quantity
            )

        print("Total:", self.calculate_total())


laptop = Product(
    "P01",
    "Laptop",
    120000,
    10
)

mouse = Product(
    "P02",
    "Mouse",
    2500,
    20
)


order = Order("Ali")
order.add_product(laptop, 2)
order.add_product(mouse, 3)

order.display_order()

print("\nRemaining Laptop Stock:", laptop.get_stock())
print("Remaining Mouse Stock:", mouse.get_stock())
```
