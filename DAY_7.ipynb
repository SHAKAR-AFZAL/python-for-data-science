```python
# SETS IN PYTHON
a = {1,2,3,4,5,6,7}
b=set()
print(a)
print(type(a))
print(type(b))
```

```python
# Hetro Sets
s = { 1,2,3.5,"4M",(6,7,8)}
print(s)
```

```python
# Homo Sets
s = { 1,2,3,1,2,6,4,56,8}
print(s)
```

```python
# DUPLICATES
s = { 1,True,0,False}
print(s)
```

```python
# Swapping Two Numbers Without Third Variable
a=10
b=20
a,b=b,a
print(a,b)
```

```python
# ADDING ELEMENT IN SETS
s = { 1,2,3,1,2,6,4,56,8}
s.add(9)
print(s)
```

```python
# ADDING ELEMENTS IN SETS
s = { 1,2,3,1,2,6,4,5,8}
s.update([9,10,11,12,13])
print(s)
```

```python
# REMOVING ELEMENTS
s = {1,2,3,1,2,6,4,5,8}
s.discard(2)
print(s)
```

```python
# Remove
s = {1,2,3,1,2,6,4,5,8}
s.remove(1)
print(s)
```

```python
# Pop
s = {1,2,3,1,2,6,4,5,8}
s.pop()
print(s)
```

```python
# Clear
s = {1,2,3,1,2,6,4,5,8}
s.clear()
print(s)
```

```python
# Set Operations
s1 = { 1,2,3,4,5}
s2 = { 4,5,6,7,8}
print(s1|s2)
print(s1&s2)
print(s1^s2)
print(1 in s1)
print(5 not in s1)
```

```python
# Set Functions
s1 = { 1,2,3,4,5}
s2 = { 4,5,6,7,8}
print(len(s1))
print(max(s2))
print(min(s1))
print(sum(s1))
print(sorted(s1))

```

```python
fs = frozenset ({1,2,3,4,5,6})
print(fs)
print(type(fs))
```
